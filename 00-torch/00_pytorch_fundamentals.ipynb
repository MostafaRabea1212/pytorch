### Reshape a tensor

### `.reshape(1, -1)`

- `.reshape()` changes the shape of the tensor without changing its values.
- `1` means the tensor will have **1 row**.
- `-1` tells PyTorch to automatically calculate the number of columns.
- Example: a tensor with `10` values becomes a shape of **`(1, 10)`**.
- You can also reshape a tensor into **3D** by providing three dimensions, such as `.reshape(1, 3, 3)`, as long as the total number of elements stays the same.

````python
tensor_f32 , tensor_f32.shape ,tensor_f32.dtype
````

````python
reshape_tensor=tensor_f32.reshape(1,10)
reshape_tensor , reshape_tensor.shape
````

````python
reshape_tensor_3d=tensor_f32.reshape(1,1,-1)
reshape_tensor_3d , reshape_tensor_3d.shape , reshape_tensor_3d.ndim
````

````python
reshape_tensor_v2=tensor_f32.reshape(2,-1)
reshape_tensor_v2
````

### `torch.stack()`

`torch.stack([x, x, x])` combines multiple tensors by adding a **new dimension**.  
Here, `x` is repeated 3 times and placed on this new dimension.  
So if `x.shape = (3,)`, the result will have `shape = (3, 3)`.  
**`stack()` increases the number of dimensions by 1.**

````python
x_stacked=torch.stack([x,x,x])
x_stacked ,x_stacked.shape
````

### `.squeeze()` and `.unsqueeze()`

- `squeeze()` removes dimensions with size `1`. For example, `(1, 3, 3)` → `(3, 3)`.
- `unsqueeze(dim=0)` adds a new dimension of size `1` at position `0`. For example, `(3, 3)` → `(1, 3, 3)`.
- So, `squeeze()` **removes** unnecessary `1` dimensions, while `unsqueeze()` **adds** a new `1` dimension.
- They can be thought of as opposite operations: `squeeze()` ↔ `unsqueeze()`.

````python
k=torch.arange(0,10)
k , k.shape
````

````python
k_mod=k.reshape(1,-1)
k_mod ,k_mod.shape
````

````python
k_mod_v2=k_mod.squeeze()
k_mod_v2 , k_mod_v2.shape
````

````python
k_mod_v3=k_mod_v2.unsqueeze(dim=0)
k_mod_v3 , k_mod_v3.shape
````

### `torch.permute()`

`torch.permute(i, (2, 0, 1))` changes the **order of the dimensions** without changing the values.  
The original shape is `(224, 228, 3)` → `(Height, Width, Channels)`.  
After `permute(2, 0, 1)`, the order becomes `(Channels, Height, Width)`.  
So the new shape is **`(3, 224, 228)`**.

````python
i = torch.rand(224,228,3)
i , i.shape
````

````python
i_permute = torch.permute(i,(2,0,1))
i_permute,i_permute.shape
````

### Tensor Indexing

`x[0][1][1]` is used to access a specific value inside a 3D tensor.  
The indices are read in order: **first dimension → second dimension → third dimension**.  
For `x.shape = (2, 2, 3)`: `0` selects the first matrix, `1` selects the second row, and `1` selects the second value.  
So `x[0][1][1]` returns **one single value** from the tensor.

````python
x=torch.rand(2,2,3)
x
````

````python
x[0][1][1]
````

### Moving Tensors Between CPU and GPU

- `torch.cuda.is_available()` → Checks if a CUDA-enabled GPU is available.
- `tensor.to(device)` → Moves the tensor to the selected device (`CPU` or `GPU`).
- `tensor.device` → Shows where the tensor is currently stored.
- `gpu_tensor.cpu()` → Moves a GPU tensor back to the CPU.
- `.cpu().numpy()` → Moves the tensor to the CPU and converts it into a NumPy array.
- `type(cpu_tensor)` → Shows that it is still a PyTorch Tensor, while `type(cpu_tensor_tensor)` shows that it is a NumPy array.
